## 00. パタトクカシーー
2つの文字列「パトカー」と「タクシー」の文字を先頭から交互に連結し、文字列「パタトクカシーー」を得よ。

```python
# prompt: 2つの文字列「パトカー」と「タクシー」の文字を先頭から交互に連結し、文字列「パタトクカシーー」を得よ。

str1 = "パトカー"
str2 = "タクシー"
result = ""

for i in range(min(len(str1), len(str2))):
  result += str1[i] + str2[i]

print(result) # Output: パタトクカシー
```

## 04. 元素記号
"Hi He Lied Because Boron Could Not Oxidize Fluorine. New Nations Might Also Sign Peace Security Clause. Arthur King Can."という文を単語に分解し、1, 5, 6, 7, 8, 9, 15, 16, 19番目の単語は先頭の1文字、それ以外の単語は先頭の2文字を取り出し、取り出した文字列から単語の位置（先頭から何番目の単語か）への連想配列（辞書型もしくはマップ型）を作成せよ。

```python
# prompt: "Hi He Lied Because Boron Could Not Oxidize Fluorine. New Nations Might Also Sign Peace Security Clause. Arthur King Can."という文を単語に分解し、1, 5, 6, 7, 8, 9, 15, 16, 19番目の単語は先頭の1文字、それ以外の単語は先頭の2文字を取り出し、取り出した文字列から単語の位置（先頭から何番目の単語か）への連想配列（辞書型もしくはマップ型）を作成せよ。

sentence = "Hi He Lied Because Boron Could Not Oxidize Fluorine. New Nations Might Also Sign Peace Security Clause. Arthur King Can."
words = sentence.replace(".", "").split()
indices = [1, 5, 6, 7, 8, 9, 15, 16, 19]
result = {}

for i, word in enumerate(words):
  if i + 1 in indices:
    result[word[:1]] = i + 1
  else:
    result[word[:2]] = i + 1

result
```

## 05. n-gram
与えられたシーケンス（文字列やリストなど）からn-gramを作る関数を作成せよ。この関数を用い、"I am an NLPer"という文から文字tri-gram、単語bi-gramを得よ。

```python
# prompt: 与えられたシーケンス（文字列やリストなど）からn-gramを作る関数を作成せよ。この関数を用い、"I am an NLPer"という文から文字tri-gram、単語bi-gramを得よ。

def n_gram(sequence, n):
  """
  与えられたシーケンスからn-gramを生成する関数。

  Args:
    sequence: 文字列またはリスト。
    n: n-gramのn。

  Returns:
    n-gramのリスト。
  """
  ngrams = []
  for i in range(len(sequence) - n + 1):
    ngrams.append(sequence[i:i + n])
  return ngrams

# 文字tri-gram
sentence = "I am an NLPer"
characters = sentence.replace(" ", "")  # スペースを除去
character_trigrams = n_gram(characters, 3)
print(f"文字tri-gram: {character_trigrams}")

# 単語bi-gram
words = sentence.split()
word_bigrams = n_gram(words, 2)
print(f"単語bi-gram: {word_bigrams}")
```

## 06. 集合
"paraparaparadise"と"paragraph"に含まれる文字bi-gramの集合を、それぞれ, $X$ と $Y$ として求め、$X$ と $Y$ の和集合（$X \cup Y$）、積集合（$X \cap Y$）、差集合（$X \setminus Y$）を求めよ。さらに、'se'というbi-gramがXおよびYに含まれるかどうかを調べよ。

```python
# prompt: "paraparaparadise"と"paragraph"に含まれる文字bi-gramの集合を、それぞれ,  𝑋  と  𝑌  として求め、 𝑋  と  𝑌  の和集合（ 𝑋∪𝑌 ）、積集合（ 𝑋∩𝑌 ）、差集合（ 𝑋∖𝑌 ）を求めよ。さらに、'se'というbi-gramがXおよびYに含まれるかどうかを調べよ。

def n_gram(sequence, n):
    ngrams = []
    for i in range(len(sequence) - n + 1):
        ngrams.append(sequence[i:i + n])
    return ngrams

X = set(n_gram("paraparaparadise", 2))
Y = set(n_gram("paragraph", 2))

union_set = X | Y
intersection_set = X & Y
difference_set = X - Y

print(f"X: {X}")
print(f"Y: {Y}")
print(f"和集合 (X ∪ Y): {union_set}")
print(f"積集合 (X ∩ Y): {intersection_set}")
print(f"差集合 (X \\ Y): {difference_set}")

print(f"'se' in X: {'se' in X}")
print(f"'se' in Y: {'se' in Y}")
```

## 07. テンプレートによる文生成
引数x, y, zを受け取り「x時のyはz」という文字列を返す関数を実装せよ。さらに、x=12, y="気温", z=22.4として、実行結果を確認せよ。

```python
# prompt: 引数x, y, zを受け取り「x時のyはz」という文字列を返す関数を実装せよ。さらに、x=12, y="気温", z=22.4として、実行結果を確認せよ。

def generate_sentence(x, y, z):
  """
  引数x, y, zを受け取り「x時のyはz」という文字列を返す関数。

  Args:
    x: 時間を表す値。
    y: 項目を表す文字列。
    z: 値を表す値。

  Returns:
    「x時のyはz」という文字列。
  """
  return f"{x}時の{y}は{z}"

# x=12, y="気温", z=22.4として実行
x = 12
y = "気温"
z = 22.4
result = generate_sentence(x, y, z)
print(result) # Output: 12時の気温は22.4
```

## 08. 暗号文
与えられた文字列の各文字を、以下の仕様で変換する関数cipherを実装せよ。

* 英小文字ならば (219 - 文字コード) のASCIIコードに対応する文字に置換
* その他の文字はそのまま出力

この関数を用い、英語のメッセージを暗号化・復号化せよ。

```python
# prompt: 与えられた文字列の各文字を、以下の仕様で変換する関数cipherを実装せよ。
# 英小文字ならば (219 - 文字コード) のASCIIコードに対応する文字に置換
# その他の文字はそのまま出力
# この関数を用い、英語のメッセージを暗号化・復号化せよ。

def cipher(text):
    result = ''
    for char in text:
        if 'a' <= char <= 'z':
            result += chr(219 - ord(char))
        else:
            result += char
    return result

# 暗号化
message = "This is a message."
encrypted_message = cipher(message)
print(f"暗号化: {encrypted_message}")

# 復号化
decrypted_message = cipher(encrypted_message)
print(f"復号化: {decrypted_message}")
```

## 09. Typoglycemia
スペースで区切られた単語列に対して、各単語の先頭と末尾の文字は残し、それ以外の文字の順序をランダムに並び替えるプログラムを作成せよ。ただし、長さが４以下の単語は並び替えないこととする。適当な英語の文（例えば"I couldn't believe that I could actually understand what I was reading : the phenomenal power of the human mind ."）を与え、その実行結果を確認せよ。

```python
# prompt: スペースで区切られた単語列に対して、各単語の先頭と末尾の文字は残し、それ以外の文字の順序をランダムに並び替えるプログラムを作成せよ。ただし、長さが４以下の単語は並び替えないこととする。適当な英語の文（例えば"I couldn't believe that I could actually understand what I was reading : the phenomenal power of the human mind ."）を与え、その実行結果を確認せよ。

import random

def typoglycemia(text):
    words = text.split()
    result = []
    for word in words:
        if len(word) <= 4:
            result.append(word)
        else:
            middle = list(word[1:-1])
            random.shuffle(middle)
            result.append(word[0] + "".join(middle) + word[-1])
    return " ".join(result)

text = "I couldn't believe that I could actually understand what I was reading : the phenomenal power of the human mind ."
print(typoglycemia(text))
```
